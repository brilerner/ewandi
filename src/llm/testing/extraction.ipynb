```python
def json_request(messages, model="gpt-4-1106-preview"):
    import json

    from openai import OpenAI

    client = OpenAI()
    # messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0,
        response_format={"type": "json_object"},
    )
    content = response.choices[0].message.content
    return json.loads(content)
```

```python
conv = Conversation()
conv.add_text_message("system", engine.system)
conv.add_text_message("system", extraction.prompt)

query = """How often have I had headache in the morning around John and Andy in the past 3 months?"""

query = """How often do I play baseball?"""
conv.add_text_message("user", query)
#
# GPT_MODEL = "gpt-3.5-turbo-1106"
GPT_MODEL = "gpt-4-1106-preview"
json_request(conv.messages, model=GPT_MODEL)
```
