```python
# foods_manual is a dict containing categories as keys and lists of foods as values
# lets convert it to a string that looks like the yaml does


def convert_foods_to_string(foods):
    foods_str = ""
    for category, foods in foods.items():
        foods_str += f"{category}:\n"
        for food in foods:
            foods_str += f"  - {food}\n"
    return foods_str


foods_manual_str = convert_foods_to_string(foods_manual)
print(foods_manual_str)
```

```python
def print_foods_by_category(foods):
    print("Number of foods: ", len(foods), "\n")
    food_by_cat = {}
    for f in foods:
        cat = f["foodCategory"]["description"]
        if cat not in food_by_cat:
            food_by_cat[cat] = []
        food_by_cat[cat].append(f["description"])

    for k, v in food_by_cat.items():
        print(k, len(v))
        for i in v:
            print(i)
        print()
```
