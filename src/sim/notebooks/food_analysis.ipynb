```python
# for each recipe type, print the frequency that ingredients occur
"""
Example of how data is stored:
{'breakfast': [{'name': 'Loaded Cheesy Beef & Potato Skillet',
   'ingredients': ['bread',
    'beef',
    'butter',
    'cheddar cheese',
"""

from collections import Counter
recipe_freqs = {}
for recipe_type, recipe_list in recipes.items():
    print(recipe_type)
    ingredients = [ing for rec in recipe_list for ing in rec['ingredients']]
    counter = Counter(ingredients)
    # divide by the number of recipes to get the frequency
    freqs = {k: v/len(recipe_list) for k, v in counter.items()}
    # sort the ingredients by frequency
    freqs = dict(sorted(freqs.items(), key=lambda item: item[1], reverse=True))
    print(freqs)
    recipe_freqs[recipe_type] = freqs
```
