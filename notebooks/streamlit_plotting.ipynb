```python
import numpy as np
import plotly.express as px

# Simulate some data
np.random.seed(42)  # For reproducibility
dates = pd.date_range("2024-01-01", periods=100)
names = ["EventA", "EventB", "EventC"]
data = {
    "start_datetime": np.tile(dates, len(names)),
    "value": np.random.rand(300),
    "name": np.repeat(names, len(dates)),
}

df = pd.DataFrame(data)

# Plotting using Plotly, ensuring each event type (name) gets a different color
fig = px.scatter(
    df, x="start_datetime", y="value", color="name", title="Simulated Event Schedule"
)

# Update layout for better visualization
fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Value",
)

fig.show()
```
