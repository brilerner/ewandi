```python

start_date = datetime(2023, 9, 1)
end_date = datetime(2023, 9, 30)
cat = 'calendar_events'
event_name = 'Basketball Game'


events = get_events(start_date, end_date, cat, event_name=event_name)


# Convert the data to a DataFrame
data = events[0]['calendar_events']
df = pd.DataFrame(data)

# Create a date range for the x-axis
start_date = df['start_datetime'].min()
end_date = df['start_datetime'].max()
date_range = pd.date_range(start=start_date, end=end_date, freq='D')

# Prepare the data for plotting
plot_data = pd.DataFrame({'date': date_range})
plot_data['event_occurrence'] = plot_data['date'].isin(df['start_datetime']).astype(int)

# Plotting using Plotly
fig = px.bar(plot_data, x='date', y='event_occurrence', title='Event Occurrence Over Time')
fig.update_layout(xaxis_title='Date', yaxis_title='Event Occurred (1=Yes, 0=No)', yaxis=dict(tickvals=[0, 1]))
fig.show()
```

```python
def plot_occurrence(variable_name):
    events = get_events(event_name=variable_name)
    return plot_events(events)

def plot_events(events):

    # Convert the data to a DataFrame
    data = events[0]['calendar_events']
    df = pd.DataFrame(data)

    # Map each event type to a unique integer
    event_type_mapping = {event: i+1 for i, event in enumerate(df['name'].unique())}
    df['event_type_id'] = df['name'].map(event_type_mapping)

    # Plotting using Plotly
    fig = px.scatter(df, x='start_datetime', y='event_type_id', color='name', title='Event Schedule')
    fig.update_layout(xaxis_title='Date', yaxis_title='Event Type', yaxis=dict(tickmode='array', tickvals=list(event_type_mapping.values()), ticktext=list(event_type_mapping.keys())))
    fig.show()
    return fig

fig = plot_occurrence('Basketball Game')
```
